```python
# ==========================================
# 作業名稱：Python 聚餐費用試算器
# 姓名：趙偉翔
# 學號：D1343778
# ==========================================

print("=== 測試一：原始資料（人數 4 人，折價券 100 元） ===")
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"
service_rate = 0.10
discount = 100
budget_per_person = 300

# 1. 整理聚餐名稱：去除頭尾空白並轉成小寫
cleaned_event_name = event_name.strip().lower()

# 2. 整理餐費與人數並轉換型別
# 移除 "NT$", 逗號與空白
cleaned_meal_amount_str = meal_amount.replace("NT$", "").replace(",", "").strip()
original_meal_amount = float(cleaned_meal_amount_str) # 轉成浮點數以便計算
total_people = int(people) # 文字轉成整數

# 3. 分步計算費用
service_fee = original_meal_amount * service_rate # 服務費 = 原始餐費 × 服務費率
total_amount = original_meal_amount + service_fee - discount # 應付總額 = 原始餐費 ＋ 服務費 － 折價券
per_person_cost = total_amount / total_people # 每人費用 = 應付總額 ÷ 分攤人數

# 4. 判斷是否超過預算
is_over_budget = per_person_cost > budget_per_person

# 5. 輸出費用摘要
print(f"聚餐名稱：{cleaned_event_name}")
print(f"原始餐費：{original_meal_amount:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{total_people} 人")
print(f"每人費用：{per_person_cost:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}\n")


print("=== 測試二：只將人數改成 '5'，折價券維持 100 元 ===")
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "5"  # 修改人數
service_rate = 0.10
discount = 100
budget_per_person = 300

cleaned_event_name = event_name.strip().lower()
original_meal_amount = float(meal_amount.replace("NT$", "").replace(",", "").strip())
total_people = int(people)

service_fee = original_meal_amount * service_rate
total_amount = original_meal_amount + service_fee - discount
per_person_cost = total_amount / total_people
is_over_budget = per_person_cost > budget_per_person

print(f"聚餐名稱：{cleaned_event_name}")
print(f"原始餐費：{original_meal_amount:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{total_people} 人")
print(f"每人費用：{per_person_cost:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}\n")


print("=== 測試三：人數恢復為 '4'，只將折價券改成 220 元 ===")
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"
service_rate = 0.10
discount = 220  # 修改折價券
budget_per_person = 300

cleaned_event_name = event_name.strip().lower()
original_meal_amount = float(meal_amount.replace("NT$", "").replace(",", "").strip())
total_people = int(people)

service_fee = original_meal_amount * service_rate
total_amount = original_meal_amount + service_fee - discount
per_person_cost = total_amount / total_people
is_over_budget = per_person_cost > budget_per_person

print(f"聚餐名稱：{cleaned_event_name}")
print(f"原始餐費：{original_meal_amount:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{total_people} 人")
print(f"每人費用：{per_person_cost:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
```
